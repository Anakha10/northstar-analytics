```python
db = client["northstar_db"]
```

```python
complaints_col = db["complaints"]
```

```python
complaint_data = {
    "customer_id": 101,
    "complaint_type": "Late Delivery",
    "city": "London",
    "status": "Open",
    "messages": [
        {"sender": "customer", "text": "My delivery was late"},
        {"sender": "support", "text": "We are checking the issue"}
    ]
}

complaints_col.insert_one(complaint_data)
```

```python
for item in complaints_col.find():
    print(item)
```

```python
complaints_col.update_one(
    {"customer_id": 101},
    {"$set": {"status": "Resolved"}}
)
```

```python
for item in complaints_col.find():
    print(item)
```

```python
complaints_col.delete_one({"customer_id": 101})
```
