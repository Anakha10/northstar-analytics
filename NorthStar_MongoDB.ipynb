```python
complaint_data = {
    "customer_id": 101,
    "complaint_type": "Late Delivery",
    "city": "London",
    "status": "Open",
    "messages": [
        {"sender": "customer", "text": "My delivery was late"},
        {"sender": "support", "text": "We are checking the issue"}
    ]
}

complaints_col.insert_one(complaint_data)
```
